###Input - Proses - Output
Input: Aksi pengguna

Proses: Disimpan ke undo stack, dipindahkan ke redo stack saat undo

Output: Riwayat aksi


###Flowchart

Start

  ↓

Input Aksi

  ↓

Push ke Undo Stack

  ↓

Pilih Aksi?

  ↓

 Undo / Redo / Tampil

  ↓

Pindah Stack /  Balik Stack /  Tampilkan Stack

  ↓
  
 End

###Pseudocode
push(data):
    tambah ke undo
    kosongkan redo

undo():
    pindah dari undo ke redo

redo():
    pindah dari redo ke undo

```python
undo_stack = []
redo_stack = []

def push(action):
    undo_stack.append(action)
    redo_stack.clear()


def undo():
    if undo_stack:
        action = undo_stack.pop()
        redo_stack.append(action)
        return action
    return "Tidak ada aksi"


def redo():
    if redo_stack:
        action = redo_stack.pop()
        undo_stack.append(action)
        return action
    return "Tidak ada aksi"


def show():
    print("Undo:", undo_stack)
    print("Redo:", redo_stack)


while True:
    print("\n1. Tambah Aksi")
    print("2. Undo")
    print("3. Redo")
    print("4. Lihat Stack")
    print("5. Keluar")

    pilih = input("Pilih: ")

    if pilih == "1":
        aksi = input("Aksi: ")
        push(aksi)
    elif pilih == "2":
        print("Undo:", undo())
    elif pilih == "3":
        print("Redo:", redo())
    elif pilih == "4":
        show()
    elif pilih == "5":
        break
```
